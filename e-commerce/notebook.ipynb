```python
import pandas as pd
import glob
```

```python
files = glob.glob("/Users/aysesudekarabal/data-cleaning/e-commerce/e-commerce-dirty-datas/*.csv")
```

```python
dfs = []
for f in files:
    df = pd.read_csv(f)
    print(df.columns)
    dfs.append(df)
combined = pd.concat(dfs,ignore_index = True)
```

```python
combined["goods-title-link--jump"]= combined["goods-title-link--jump"].fillna(combined["goods-title-link"])
```

```python
combined["goods-title-link--jump"].isna().sum()
```

```python
combined["goods-title-link--jump"]= combined["goods-title-link--jump"].fillna("Unknown")
```

```python
combined = combined.rename(columns= {"goods-title-link--jump":"title"})
```

```python
combined = combined.drop(columns = ["goods-title-link"])
```

```python
combined.head()
```

```python
combined["price"] = combined["price"].str.replace('$',' ')
```

```python
combined["discount"] = combined["discount"].str.replace('%', ' ')
```

```python
combined.head()
```

```python
combined["price"] = pd.to_numeric(combined["price"],errors = "coerce")
```

```python
combined["discount"] = pd.to_numeric(combined["discount"],errors = "coerce")
combined["color-count"] = pd.to_numeric(combined["color-count"],errors = "coerce")
```

```python
combined.dtypes
```

```python
combined.duplicated(subset = ["title","price"]).sum()
```

```python
combined = combined.drop_duplicates(subset = ["title","price"])
```

```python
# 1. Sayıyı ve ardından gelen 'k' veya 'm' harfini ayır
extracted = combined["selling_proposition"].str.extract(r'([\d.]+)([kKmM]?)')

# 2. Sayısal kısmı dönüştür
nums = pd.to_numeric(extracted[0], errors='coerce')

# 3. Harflere göre çarpanları belirle (k = 1.000, m = 1.000.000)
multipliers = extracted[1].str.lower().map({'k': 1000, 'm': 1000000}).fillna(1)

# 4. Çarpıp yeni sütuna yaz
combined["sold_over_count"] = nums * multipliers
```
